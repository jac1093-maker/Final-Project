The following code is to provide the amount of points each neighborhood won based on the amount of water features it has and its type or characteristics from a made up point system with the purpose to quantify the data for comparison. The made up point system gives credit where it is needed. The system gives 16 points if a water feature is an active drinking fountain, 13 points if a water feature is an active spray, 10 points if a water feature is an active decorative, and 10 points if a water feature is an active pond. The reason why a drinking fountain received the most points is because drinking water is a necessity and serves as a function of need for passerbys who need it. The reason why the spray received the second most points is because a spray helps cool down and provide moisture outside where there is no air conditioning especially during hot days with little humidity which is also quite important as a need, but isn't as important as drinking. The reason why the decorative received the lowest points is because it is for aesthetics or looks rather than a purpose of necessity. The pond got the same score because its function is unknown, but it still provides natural aesthetics like the decorative which is why it received the same points. However, it does bring attraction compared to the other two types of water features. The reason why the points are only separated by three is because the main focus or goal is still on the quantity or amount of active features a neighborhood has to determine the best neighborhood. Additionally, the types of water feature still serve a specific purpose based on what it is making it important. However, a water feature would only receive two points if it is inactive regardless of its type because there was credit given for having the structure built, but they all won't provide any functional purpose due to its inactivity. On top of the points received based on the type and whether it is active, the water feature also gains additional points based on its control type. The drinking fountain gains one point for having an on/off control type and zero points for having a continuous control type because the on/off control type prevents wasting water that no one drinks. The spray also gains one point for having an on/off control type and zero points for having a continuous control type for the same reason except it prevents wasting water that no one uses to cool off. On the other hand, the decorative receives zero points for having a on/off control type, while one point for having a continuous control type. The reason is because a decorative reuses the water by looping it based on its design which means barely any water would be wasted from a continuous control type and it looks better for decorative than an on/off control type. Finally, the pond recevies an additional of zero points no matter what because the control type doesn't apply to it. There was a gain or loss of three points for the type of water feature because the type is important to determine its function. On the other hand, the inactivity of the water feature received two points to give it some credit without affecting its overall score too much. Finally, only one or zero additional points were given for the control type because the how it functions part isn't as significant as what the function is including that there are pros and cons to both control types. Thus, the points were made smaller to not impact the ranking as much, but can still make a difference if there is enough of it. 

```python
water_features["active_inactive_score"] = 0 

for index, row in water_features.iterrows():
    if row["feature_type"] == "Drinking Fountain":
        if row["inactive"] == False:
            water_features.loc[index, "active_inactive_score"] = 16 
        elif row["inactive"] == True:
            water_features.loc[index, "active_inactive_score"] = 2

    elif row["feature_type"] == "Spray":
        if row["inactive"] == False:
            water_features.loc[index, "active_inactive_score"] = 13 
        elif row["inactive"] == True:
            water_features.loc[index, "active_inactive_score"] = 2 

    elif row["feature_type"] == "Decorative":
        if row["inactive"] == False:
            water_features.loc[index, "active_inactive_score"] = 10 
        elif row["inactive"] == True:
            water_features.loc[index, "active_inactive_score"] = 2

    elif row["feature_type"] == "Pond":
        if row["inactive"] == False:
            water_features.loc[index, "active_inactive_score"] = 10 
        elif row["inactive"] == True:
            water_features.loc[index, "active_inactive_score"] = 2

water_features["control_type_score"] = 0 

for index, row in water_features.iterrows():
    if row["feature_type"] == "Drinking Fountain":
        if row["control_type"] == "On/Off":
            water_features.loc[index, "control_type_score"] = 1 
        elif row["control_type"] == "Continuous": 
            water_features.loc[index, "control_type_score"] = 0

    elif row["feature_type"] == "Spray":
        if row["control_type"] == "On/Off":
            water_features.loc[index, "control_type_score"] = 1
        elif row["control_type"] == "Continuous":
            water_features.loc[index, "control_type_score"] = 0 

    elif row["feature_type"] == "Decorative":
        if row["control_type"] == "On/Off":
            water_features.loc[index, "control_type_score"] = 0 
        elif row["control_type"] == "Continuous":
            water_features.loc[index, "control_type_score"] = 1

    elif row["feature_type"] == "Pond":
        water_features.loc[index, "control_type_score"] = 0

water_features["score"] = (water_features["active_inactive_score"] + water_features["control_type_score"]) 

pittsburgh_neighborhood_score = water_features.groupby("neighborhood")["score"].sum() 
display(pittsburgh_neighborhood_score[0:34]) 
display(pittsburgh_neighborhood_score[34:])
```

The purpose of the code below is to rearrange the neighborhoods below into numerical descending order based on the score received to determine which neighborhood scored the highest and is the best in Pittsburgh. The reason for displaying the results twice is because it doesn't display all the scores due to having a lot of neighborhoods. 

```python
pittsburgh_neighborhood_score_sort = pittsburgh_neighborhood_score.sort_values(ascending=False)
display(pittsburgh_neighborhood_score_sort[0:34])
display(pittsburgh_neighborhood_score_sort[34:]) 
```

The purpose of the code below is to display the rank of each neighborhood in numerical ascending order. The neighborhoods that have the same score receive the same rank for fairness rather than by alphabetically. 

```python
pittsburgh_neighborhood_score_rank = pittsburgh_neighborhood_score.rank(ascending=False, method='dense')
pittsburgh_neighborhood_score_rank_sort = pittsburgh_neighborhood_score_rank.sort_values(ascending=True)
display(pittsburgh_neighborhood_score_rank_sort[0:34])
display(pittsburgh_neighborhood_score_rank_sort[34:]) 
```
